 3.Count the frequency of elements using a dictionary.

```python
words =["apple", "banana", "apple", "orange", "banana", "apple"]

freq ={}

for word in words:
   freq[word] = freq.get(word, 0) + 1

   if __name__ == '__main__':
    print("Word freqencies:", freq)
```

4.Use get() and setdefault() for safe key access.

```python
user = {"name": "Carol"}
age = user.get("age", 30)
user.setdefault("city", "London")

if __name__ == '__main__':
  print("User Dictory:", user)
  print ("Age (defalt if missing):", age)
```

5.Demonstrate dictionary comprehensions.

```python
squares = {x: x*x for x in range(6)}
if __name__ == '__main__':
  print("Square Dictionary:", squares)
```

6.Show different ways to iterate through a dictionary.

```python
student = {"name": "Bob", "age": 22, "grade": "A"}

if __name__ == '__main__':
    print("Keys:")
    for key in student:
        print(key)
    print("\nValues:")
    for value in student.values():
        print(value)
    print("\nItems:")
    for key, value in student.items():
        print(f"{key} => {value}")
```

7.Invert keys and values of a dictionary.


```python
mapping = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in mapping.items()}

if __name__ == '__main__':
    print("Original:", mapping)
    print("Inverted:", inverted)
```

8.Combine two dictionaries using unpacking and update methods.

```python
a = {"x": 1, "y": 2}
b = {"y": 3, "z": 4}

merged = {**a, **b}

if __name__ == '__main__':
  print("Merged dictionary:", merged)
```

 9.A dictionary containing nested dictionaries.

```python
company = {
    "HR": {"employees": 5, "manager": "Sarah"},
    "IT": {"employees": 10, "manager": "John"},
    "Finance": {"employees": 3, "manager": "David"}
}

if __name__ == '__main__':
    for dept, info in company.items():
        print(f"{dept} Department -> Manager: {info['manager']}, Employees: {info['employees']}")
```

10.Sort a dictionary by keys and values.

```python
data = {"b": 3, "a": 5, "c":1}
sorted_by_key = dict(sorted(data.items()))
sorted_by_value = dict(sorted(data.items(), key=lambda x: x[1]))

if __name__ == '__main__':
  print("Original:", data)
  print("Sorted by Key:", sorted_by_key)
  print ("Sorted by value:", sorted_by_value)
```
