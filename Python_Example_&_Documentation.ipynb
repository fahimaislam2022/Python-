# **Types of Data Types in Python with Examples and Explanation:**

**Numeric Data Types**

Python has **three types** of numeric data types:


*  int (Integer)
*  float (Floating point number)
*  complex (Complex number)








```python
# Integer (int) example
num1 = 10  # Whole number
print(num1, type(num1))  # Output: 10 <class 'int'>

# Floating point (float) example
num2 = 10.5  # Decimal number
print(num2, type(num2))  # Output: 10.5 <class 'float'>

# Complex number (complex) example
num3 = 3 + 4j  # Complex number
print(num3, type(num3))  # Output: (3+4j) <class 'complex'>
```

**Boolean Data Type**

```python
is_active = True
is_logged_in = False
print(is_active, type(is_active))  # Output: True <class 'bool'>
print(is_logged_in, type(is_logged_in))  # Output: False <class 'bool'>
```

**String Data Type**

```python
# String example
text1 = "Hello, World!"  # Double quotes
text2 = 'Python Programming'  # Single quotes
text3 = """This is a
multi-line string."""  # Triple quotes

print(text1, type(text1))  # Output: Hello, World! <class 'str'>
print(text2, type(text2))  # Output: Python Programming <class 'str'>
print(text3, type(text3))  # Output: This is a multi-line string. <class 'str'>
```

# **Operators in Python with Examples**

Python provides various types of operators. Here, we'll cover:
*   Arithmetic Operators
*   Assignment Operators
*   Comparison Operators







```python
a = 10
b = 3

# Addition (+)
print("Addition:", a + b)  # Output: 13

# Subtraction (-)
print("Subtraction:", a - b)  # Output: 7

# Multiplication (*)
print("Multiplication:", a * b)  # Output: 30

# Division (/)
print("Division:", a / b)  # Output: 3.3333 (float division)

# Floor Division (//)
print("Floor Division:", a // b)  # Output: 3 (removes decimal part)

# Modulus (%)
print("Modulus:", a % b)  # Output: 1 (remainder of division)

# Exponentiation (**)
print("Exponentiation:", a ** b)  # Output: 1000 (10^3)
```

```python
x = 5  # Simple assignment
print("x:", x)  # Output: 5

x += 3  # x = x + 3
print("x += 3:", x)  # Output: 8

x -= 2  # x = x - 2
print("x -= 2:", x)  # Output: 6

x *= 4  # x = x * 4
print("x *= 4:", x)  # Output: 24

x /= 3  # x = x / 3
print("x /= 3:", x)  # Output: 8.0
```

```python
x = 10
y = 5

print("x == y:", x == y)  # False (10 is not equal to 5)
print("x != y:", x != y)  # True (10 is not equal to 5)
print("x > y:", x > y)  # True (10 is greater than 5)
print("x < y:", x < y)  # False (10 is not less than 5)
print("x >= y:", x >= y)  # True (10 is greater than or equal to 5)
print("x <= y:", x <= y)  # False (10 is not less than or equal to 5)
```

# **Type Casting**
Type casting (also called type conversion) is the process of converting one data type into another.
**Python supports both implicit (automatic) and explicit (manual) type casting.**




**Implicit Type Casting (Automatic Conversion)**

```python
a = 5       # Integer
b = 2.5     # Float
c = a + b   # Python converts 'a' to float automatically

print(c)    # Output: 7.5
print(type(c))  # Output: <class 'float'>
```

**Explicit Type Casting (Manual Conversion)**

```python
num = 10    # Integer
new_num = float(num)  # Explicitly converting int to float

print(new_num)    # Output: 10.0
print(type(new_num))  # Output: <class 'float'>


num_str = "100"  # String
num = int(num_str)  # Converts string to integer

print(num + 50)  # Output: 150
print(type(num))  # Output: <class 'int'>
```

# **For Loop**

A for loop is used when we know the number of iterations beforehand. It iterates over a sequence (list, tuple, string, etc.).

```python
fruits = ["Apple", "Banana", "Cherry"]
for i in fruits:
    print(i)
```

# **While Loop**
A while loop is used when the number of iterations is not known beforehand and depends on a condition.

```python
num = 1
while num <= 5:
    print(num)
    num += 1  # Increment num
```

```python
password = ""
while password != "admin123":
    password = input("Enter password: ")
print("Access granted!")
```

# **Break & Continue**

```python
for num in range(1, 10):
    if num == 5:
        print("Point", num)
        break  # Exits the loop when num is 5
    print(num)
```

```python
for num in range(1, 6):
    if num == 3:
        print("Skipping", num)
        continue  # Skips the rest of the loop body for num = 3
    print(num)
```

Characteristics of Tuples:


1.   Elements have a fixed order.

4.   Since they are immutable, tuples are optimized for speed.

# **Dictionary Functions in Python**

# **Functions in Python**
A function in Python is a block of reusable code that performs a specific task. Functions help in code organization, reusability, and modularity.
Here are two main types of functions in Python:

**Built-in Functions:** Predefined functions like print(), len(), sum(), etc.
**User-Defined Functions**: Functions created by the programmer using the def keyword

```python
def greet():
    print("Hello, welcome to Python!")
greet()
```

```python
def greet_user(name):
    print("Hello",name)
greet_user("Alice")
```

```python
def add(a, b):
    return a + b

result = add(5, 3)
print("Sum:", result)
```

```python
def greet(name="Guest"):
    print("Hello",name)
greet()          # Uses default value
greet("John")    # Uses provided value
```

```python
#Recursive Functions
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

**Function Scope (Local and Global Variables)**
**Local Variable:** Defined inside a function and accessible only within that function.
**Global Variable:** Defined outside a function and accessible anywhere.

```python
x = 10  # Global variable

def func():
    x = 5  # Local variable
    print("Inside function:", x)

func()
print("Outside function:", x)
```
