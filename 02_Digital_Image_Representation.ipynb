```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

#1 Loading and Matrix Inspection

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
print("---Example 1 : Loading Image ----")
print("Matrix Shape (Height, Width, Channels):", image_rgb.shape)
print("Data Type:", image_rgb.dtype)
print("Total Elements (Pixels x Channels):", image_rgb.size)
plt.imshow(image_rgb)
plt.title("1. Original RGB Image")
plt.axis('off')
plt.show()
```

#2 Extracting the Red Channel Matrix  

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
red_channel = image_rgb[:, :, 0]
print("---Example 2 : Red Channel ----")
print("Red Matrix Shape:", red_channel.shape)
print("Maximum Red Intensity:", np.max(red_channel))
print("Minimum Red Intensity:", np.min(red_channel))
plt.imshow(red_channel, cmap='Reds')
plt.title("2. Red Channel")
plt.axis('off')
plt.show()
```

#3 Extrcting the Green Channel Matrux

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
green_channel = image_rgb[:, :, 1]
print("---Example 3: Green Channel ----")
print("Green Matrix Shape:", green_channel.shape)
print("Average Green Intensity:", np.mean(green_channel))
plt.imshow(green_channel, cmap='Greens')
plt.title("3. Green Channel")
plt.axis('off')
plt.show()
```

#4. Extracting the Blue Channel Matrix

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
blue_channel = image_rgb[:, :, 2]
print("---Example 4: Blue Channel ----")
print("Blue Matrix Shape:", blue_channel.shape)
print("Variance Blue Channel:", np.var(blue_channel))
plt.imshow(blue_channel, cmap='Blues')
plt.title("4. Blue Channel")
plt.axis('off')
plt.show()
```

#5. Manual Grayscale Conversion (Average Method)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
gray_avg = np.mean(image_rgb, axis=2).astype(np.uint8)
print("---Example 5: Manual Grayscale Conversion (Average Method) ----")
print("New Matrix Shape:", gray_avg.shape)
print("Sample Pixel Value at (10, 10):", gray_avg[10, 10])
plt.imshow(gray_avg, cmap='gray')
plt.title("5. Grayscale (Average)")
plt.axis('off')
plt.show()
```

#6. Manual Grayscale Conversion (Luminosity Method)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
red = image_rgb[:, :, 0]
green = image_rgb[:, :, 1]
blue = image_rgb[:, :, 2]
gray_lum = (0.299 * red + 0.587 * green + 0.114 * blue).astype(np.uint8)
print("Luminosity Matrix Shape:", gray_lum.shape)
print("Unique Intensity Values Count:", len(np.unique(gray_lum)))
plt.imshow(gray_lum, cmap='gray')
plt.title("6. Grayscale (Luminosity)")
plt.axis('on')
plt.show()
```

#7. Image Binarization (Thresholding)

```python
image_bgr = cv2.imread('cat.png')
gray_image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
_, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
print("Binary Matrix Shape:", binary_image.shape)
print("Unique Values in Matrix:", np.unique(binary_image))
plt.imshow(binary_image, cmap='gray')
plt.title("7. Binary Image (Threshold=127)")
plt.axis('on')
plt.show()
```

#8.Array Slicing (Image Cropping)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
cropped_image = image_rgb[50:200, 100:300]
print("Original Shape:", image_rgb.shape)
print("Cropped Shape:", cropped_image.shape)
plt.imshow(cropped_image)
plt.title("8.Cropped Image")
plt.axis('on')
plt.show()
```

#9.Matrix Resizing (Interpolation)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
resized_image = cv2.resize(image_rgb, (100,100))
print("Resized Shape:", resized_image.shape)
print("Toatal Pixels after Resize:", resized_image.shape[0] * resized_image.shape[1])
plt.imshow(resized_image)
plt.title("9. Resized Image (100x100)")
plt.axis('on')
plt.show()
```

#10. Array Transposition (Swapping Axes)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
transposed_image = np.transpose(image_rgb, (1, 0, 2))
print("Original Shape:", image_rgb.shape)
print("Transposed Shape:", transposed_image.shape)
plt.imshow(transposed_image)
plt.title("10. Transposed Image")
plt.axis('on')
plt.show()
```

#11. Color Matrix Inversion (Negative Image)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
inverted_image = 255 - image_rgb
print("Original Pixel (0,0):", image_rgb[0, 0])
print("Inverted Pixel (0,0):", inverted_image[0, 0])
plt.imshow(inverted_image)
plt.title("11. Inverted Colors")
plt.axis('on')
plt.show()
```

#12. Scalar Addition (Increasing Brightness)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
bright_image =cv2.add(image_rgb, np.array([50.0]))
print("Original Mean:", np.mean(image_rgb))
print("Brightened Mean:", np.mean(bright_image))
plt.imshow(bright_image)
plt.title("12. Increased Brightness")
plt.axis('on')
plt.show()
```

#13 Scalar Subtraction (Decreasing Brightness)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
dark_image = cv2.subtract(image_rgb, np.array([50.0]))
print("Original Mean:", np.mean(image_rgb))
print("Darkened Mean:", np.mean(dark_image))
plt.imshow(dark_image)
plt.title("13.Decreased Brightness")
plt.axis('on')
plt.show()
```

#14. Normalization (0.0 to 1.0 Floating Point)

```python
image_bgr = cv2.imread('cat.png')
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
normalized_image = image_rgb.astype(np.float32) / 255.0
print("Normalized Data Type:", normalized_image.dtype)
print("Max Value:", np.max(normalized_image))
print("Min Value:", np.min(normalized_image))
plt.imshow(normalized_image)
plt.title("14. Normalized Image")
plt.axis('on')
plt.show()
```

#15. Tensor Flattening (1D Vector Representation)

```python
image_bgr = cv2.imread('cat.png')
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
flattened_vector = image_gray.flatten()
print("2D Matrix Shape:", image_gray.shape)
print("1D Vector Shape:", flattened_vector.shape)
print("First 10 Elements:", flattened_vector[:10])
plt.plot(flattened_vector[:1000])
plt.title("15. First 1000 Flattened Pixels Plot")
plt.show()
```
