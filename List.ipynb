1.A list of coordinate pairs and a helper to compute centroids.

```python
coords = [(0,0), (2,2), (4,0), (2, -2)]

def centroid(points):
    if not points:
        return None
    xs = [p[0] for p in points]
    ys = [p[1] for p in points]
    return (sum(xs)/len(xs), sum(ys)/len(ys))

if __name__ == '__main__':
    print("Coordinates:", coords)
    print("Centroid:", centroid(coords))
```

2.Merge lists of dictionaries by concatenating list-values and updating scalars.

```python
list_a = [{"id":1, "tags":["a","b"], "count":2}, {"id":2, "tags":["c"], "count":1}]
list_b = [{"id":1, "tags":["d"], "count":3}, {"id":3, "tags":["e"], "count":4}]

def deep_merge_by_id(a, b):
    merged = {}
    for d in a + b:
        _id = d.get("id")
        if _id not in merged:
            merged[_id] = {**d}
            merged[_id].setdefault("tags", list(merged[_id].get("tags", [])))
        else:
            merged[_id]["tags"] = merged[_id].get("tags", []) + d.get("tags", [])
            if "count" in d or "count" in merged[_id]:
                merged[_id]["count"] = merged[_id].get("count", 0) + d.get("count", 0)
    return list(merged.values())

if __name__ == '__main__':
    print("List A:", list_a)
    print("List B:", list_b)
    print("Merged:", deep_merge_by_id(list_a, list_b))
```

3.Utility to flatten arbitrarily nested lists (handles tuples too).

```python
def flatten_iterable(it):
    out = []
    for x in it:
        if isinstance(x, (list, tuple)):
            out.extend(flatten_iterable(x))
        else:
            out.append(x)
    return out

example = [1, (2, 3), [4, [5, 6], (7,)], 8]

if __name__ == '__main__':
    print("Example:", example)
    print("Flattened:", flatten_iterable(example))
```

 7.A nested list and a small recursive pretty-printer.


```python
nested_list = [1, [2, [3, 4], 5], ["a", ["b", "c"]]]

def flatten(lst):
    out = []
    for item in lst:
        if isinstance(item, list):
            out.extend(flatten(item))
        else:
            out.append(item)
    return out

if __name__ == '__main__':
    print("Nested:", nested_list)
    print("Flattened:", flatten(nested_list))
```

10.Create a list with duplicates and show how to get unique values preserving order.

```python
dup_list = [3, 1, 4, 1, 5, 9, 3, 5]

def unique_preserve_order(lst):
    seen = set()
    out = []
    for x in lst:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out

if __name__ == '__main__':
    print("Original:", dup_list)
    print("Unique (order preserved):", unique_preserve_order(dup_list))
```
